# 01 — Análise exploratória das câmeras do Smart Sampa

Este notebook explora a fotografia de setembro de 2025 por subprefeitura. Nesta etapa, a análise é descritiva; não há inferência causal sobre criminalidade.

```python
from pathlib import Path
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DB = ROOT / 'database' / 'smart_sampa.sqlite'
con = sqlite3.connect(DB)
df = pd.read_sql('SELECT * FROM vw_subpref_cameras_2025_09', con)
con.close()
df.head()
```

## Checagens básicas

A soma da fotografia das 32 subprefeituras deve ser 40.000 câmeras.

```python
assert len(df) == 32
assert df['cameras_2025_09'].sum() == 40000
print('Subprefeituras:', len(df))
print('Total de câmeras:', f"{df['cameras_2025_09'].sum():,}".replace(',', '.'))
```

## Ranking absoluto

```python
df[['subprefeitura','cameras_2025_09']].sort_values('cameras_2025_09', ascending=False).head(10)
```

```python
top = df.nlargest(10, 'cameras_2025_09').sort_values('cameras_2025_09')
fig, ax = plt.subplots(figsize=(9, 5.5))
ax.barh(top['subprefeitura'], top['cameras_2025_09'])
ax.set_title('Smart Sampa: 10 subprefeituras com mais câmeras')
ax.set_xlabel('Câmeras conectadas — setembro de 2025')
ax.set_ylabel('')
plt.tight_layout();
```

## Cobertura per capita

O denominador é a população do Censo 2022. A diferença temporal é mantida explícita no nome do indicador.

```python
df[['subprefeitura','cameras_2025_09','populacao_2022','cameras_por_10_mil_hab_pop2022']].sort_values('cameras_por_10_mil_hab_pop2022', ascending=False).head(10)
```

## Densidade espacial

A área territorial é a área oficial de 2025. Esse indicador não captura população flutuante nem uso comercial do território.

```python
df[['subprefeitura','area_km2','cameras_2025_09','cameras_por_km2_area2025']].sort_values('cameras_por_km2_area2025', ascending=False).head(10)
```
